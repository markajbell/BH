```python
#Driver Code Starts{
import heapq
import sys

# Function to construct adjacency
def constructAdj(edges, V, node_id_to_index):

    # adj[u] = list of [v, wt]
    adj = [[] for _ in range(V)]

    for edge in edges:
        u, v, wt = edge
        # Convert u and v to integer indices using the mapping
        u_index = node_id_to_index[u]
        v_index = node_id_to_index[v]
        adj[u_index].append([v_index, wt])
        # Assuming the graph is undirected, add the reverse edge as well
        adj[v_index].append([u_index, wt])

    return adj

#Driver Code Ends }
# Returns shortest distances from src to all other vertices
def dijkstra(V, edges, src, node_id_to_index):
    # Create adjacency list
    adj = constructAdj(edges, V, node_id_to_index)

    # Create a priority queue to store vertices that
    # are being preprocessed.
    pq = []

    # Create a list for distances and initialize all
    # distances as infinite
    dist = [sys.maxsize] * V

    # Convert src to its integer index using the mapping
    src_index = node_id_to_index[src]

    # Insert source itself in priority queue and initialize
    # its distance as 0.
    heapq.heappush(pq, [0, src_index])
    dist[src_index] = 0

    # Looping till priority queue becomes empty (or all
    # distances are not finalized)
    while pq:
        # The first vertex in pair is the minimum distance
        # vertex, extract it from priority queue.
        u = heapq.heappop(pq)[1]

        # Get all adjacent of u.
        for x in adj[u]:
            # Get vertex label and weight of current
            # adjacent of u.
            v, weight = x[0], x[1]

            # If there is shorter path to v through u.
            if dist[v] > dist[u] + weight:
                # Updating distance of v
                dist[v] = dist[u] + weight
                heapq.heappush(pq, [dist[v], v])
                print(pq)
                #print

    # Return the shortest distance array
    return dist
#Driver Code Starts{

# Driver program to test methods of graph class
if __name__ == "__main__":
    node_count = len(df['data']['nodes'])
    V = node_count

    # Create a mapping from original node IDs to integer indices
    # Get the keys (node IDs) from the 'nodes' dictionary
    node_ids = list(df['data']['nodes'].keys())
    node_id_to_index = {node_id: index for index, node_id in enumerate(node_ids)}
    # You might want a reverse mapping if you need to output results with original IDs
    # index_to_node_id = {index: node_id for node_id, index in node_id_to_index.items()}

    i = 0
    # Construct the complete list of edges with source, target (as original IDs), and weight
    # The constructAdj function will handle the mapping to indices
    edges = [(Source[i], Target[i], Weight[i]) for i in range(len(Source))]

    # Get the source node ID (it's already an ID from Source[0])
    src_id = Source[0]

    # Call dijkstra with the node ID to index mapping
    result = dijkstra(V, edges, src_id, node_id_to_index)

    # Print shortest distances in one line
    print(' '.join(map(str, result)))

#Driver Code Ends }
```

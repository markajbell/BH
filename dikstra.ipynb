```python
import heapq
import sys

def dijkstra(graph, start_node):
  """
  Implements Dijkstra's algorithm to find the shortest paths from a start node.

  Args:
    graph: A dictionary representing the graph. Keys are nodes, and values are
           dictionaries where keys are neighboring nodes and values are edge weights.
    start_node: The node to start the algorithm from.

  Returns:
    A dictionary where keys are nodes and values are the shortest distance
    from the start_node to that node. Returns float('inf') if a node is unreachable.
  """
  # Initialize distances with infinity for all nodes except the start node
  distances = {node: float('inf') for node in graph}
  distances[start_node] = 0

  # Use a priority queue (min-heap) to store nodes to visit
  # The heap will store tuples of (distance, node)
  priority_queue = [(0, start_node)]

  while priority_queue:
    # Get the node with the smallest distance from the priority queue
    # Corrected: Remove [1] as heappop returns the tuple directly
    current_distance, current_node = heapq.heappop(priority_queue)

    # If the current distance is greater than the recorded distance,
    # we've found a shorter path already, so skip
    if current_distance > distances[current_node]:
      continue

    # Explore neighbors of the current node
    if current_node in graph:
      for neighbor, weight in graph[current_node].items():
        distance = current_distance + weight

        # If a shorter path to the neighbor is found, update the distance
        if distance < distances[neighbor]:
          distances[neighbor] = distance
          heapq.heappush(priority_queue, (distance, neighbor))

  return distances

# Example usage with the loaded data
# First, let's build the graph representation from the edges
graph = {}

# Populate the graph dictionary from the new_edges_df DataFrame
# Use iterrows() to iterate through the rows of the DataFrame
for index, row in new_edges_df.iterrows():
  source_node = row['source']
  target_node = row['target']
  # Access the 'Weight' column from the row, defaulting to 1 if not present (though you've already handled NaNs)
  weight = row['Weight'] if 'Weight' in row else 1

  # Ensure the nodes exist in the graph dictionary
  if source_node not in graph:
    graph[source_node] = {}
  if target_node not in graph:
    graph[target_node] = {}

  # Add the edge to the graph (assuming undirected for simplicity, adjust if directed)
  # For a directed graph, only add graph[source_node][target_node] = weight
  graph[source_node][target_node] = weight
  graph[target_node][source_node] = weight


# Let's pick a random node as the starting point for Dijkstra's
# You might want to choose a more meaningful starting node based on your data
if graph:
  # Ensure start_node is one of the keys in the graph
  if list(graph.keys()):
    start_node = list(graph.keys())[0] # Get the first node in the graph
    shortest_distances = dijkstra(graph, start_node)

    print(f"\nShortest distances from node '{start_node}':")
    for node, distance in shortest_distances.items():
      # Handle unreachable nodes (distance remains infinity)
      if distance == float('inf'): # Use float('inf') for comparison
          print(f"To node {node}: Unreachable")
      else:
          print(f"To node {node}: {distance}")
  else:
      print("\nGraph is empty. Cannot run Dijkstra's algorithm.")

else:
  print("\nGraph is empty. Cannot run Dijkstra's algorithm.")
  shortest_distances
```

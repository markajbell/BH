```python
import pandas as pd
import json

# Load the JSON files
with open('vul1k_nodes.json', 'r') as f:
    vul1k_data = json.load(f)

with open('vul5k_nodes.json', 'r') as f:
    vul5k_data = json.load(f)

with open('secure1k_nodes.json', 'r') as f:
    sec1k_data = json.load(f)

with open('secure10k_nodes.json', 'r') as f:
    sec10k_data = json.load(f)

with open('secure50k_nodes.json', 'r') as f:
    sec50k_data = json.load(f)


# Convert JSON data to DataFrames
df_vul1k = pd.DataFrame(list(vul1k_data.items()), columns=['Label', 'vul1k'])
df_vul5k = pd.DataFrame(list(vul5k_data.items()), columns=['Label', 'vul5k'])
df_sec1k = pd.DataFrame(list(sec1k_data.items()), columns=['Label', 'sec1k'])
df_sec10k = pd.DataFrame(list(sec10k_data.items()), columns=['Label', 'sec10k'])
df_sec50k = pd.DataFrame(list(sec50k_data.items()), columns=['Label', 'sec50k'])
#df_vul5k = pd.DataFrame(list(vul5k_data.items()), columns=['Label', 'vul5k'])

# Merge the DataFrames on the 'Label' column
merged_df = pd.merge(df_vul1k, df_vul5k, on='Label', how='outer')
merged_df = pd.merge(merged_df, df_sec1k, on='Label', how='outer')
merged_df = pd.merge(merged_df, df_sec10k, on='Label', how='outer')
merged_df = pd.merge(merged_df, df_sec50k, on='Label', how='outer').fillna(0)


# Sort the DataFrame by 'vul1k_Count' in ascending order
merged_df_sorted = merged_df.sort_values(by='vul1k', ascending=True)

# Display the sorted DataFrame
display(merged_df_sorted)
```

```python
import json
import pandas as pd

# Load the JSON files
with open('vul1k_label.json', 'r') as f:
    vul1k_label_data = json.load(f)

with open('vul5k_label.json', 'r') as f:
    vul5k_label_data = json.load(f)

# Convert JSON data to DataFrames
df_vul1k_label = pd.DataFrame(list(vul1k_label_data.items()), columns=['Label', 'vul1k_Label_Count'])
df_vul5k_label = pd.DataFrame(list(vul5k_label_data.items()), columns=['Label', 'vul5k_Label_Count'])

# Merge the DataFrames on the 'Label' column
merged_label_df = pd.merge(df_vul1k_label, df_vul5k_label, on='Label', how='outer').fillna(0)

# Sort the DataFrame by 'vul1k_Label_Count' in ascending order
merged_label_df_sorted = merged_label_df.sort_values(by='vul1k_Label_Count', ascending=True)

# Display the merged and sorted DataFrame
display(merged_label_df_sorted)
```

```python
import pandas as pd
import json

# Load the JSON files
with open('vul1k_label.json', 'r') as f:
    vul1k_data = json.load(f)

with open('vul5k_label.json', 'r') as f:
    vul5k_data = json.load(f)

with open('secure1k_label.json', 'r') as f:
    sec1k_data = json.load(f)

with open('secure10k_label.json', 'r') as f:
    sec10k_data = json.load(f)

with open('secure50k_label.json', 'r') as f:
    sec50k_data = json.load(f)


# Convert JSON data to DataFrames
df_vul1k = pd.DataFrame(list(vul1k_data.items()), columns=['Label', 'vul1k_label'])
df_vul5k = pd.DataFrame(list(vul5k_data.items()), columns=['Label', 'vul5k_label'])
df_sec1k = pd.DataFrame(list(sec1k_data.items()), columns=['Label', 'sec1k_label'])
df_sec10k = pd.DataFrame(list(sec10k_data.items()), columns=['Label', 'sec10k_label'])
df_sec50k = pd.DataFrame(list(sec50k_data.items()), columns=['Label', 'sec50k_label'])


# Merge the DataFrames on the 'Label' column
merged_df = pd.merge(df_vul1k, df_vul5k, on='Label', how='outer')
merged_df = pd.merge(merged_df, df_sec1k, on='Label', how='outer')
merged_df = pd.merge(merged_df, df_sec10k, on='Label', how='outer')
merged_df = pd.merge(merged_df, df_sec50k, on='Label', how='outer').fillna(0)


# Sort the DataFrame by 'vul1k_label' in ascending order
merged_df_sorted = merged_df.sort_values(by='vul1k_label', ascending=True)

# Display the sorted DataFrame
display(merged_df_sorted)
```

```python
#

import pandas as pd
import heapq

def dijkstra_shortest_path(edges, nodes, start_node_id):
    """
    Finds the shortest paths from a start node to all other nodes
    using Dijkstra's algorithm.

    Args:
        edges (pd.DataFrame): DataFrame with 'source', 'target', and 'weight' columns.
        nodes (pd.DataFrame): DataFrame with 'node_id'.
        start_node_id (int): The ID of the starting node.

    Returns:
        dict: A dictionary where keys are the reachable node IDs and values are their shortest distances.
        dict: A dictionary to reconstruct the shortest paths.
    """
    # Build the graph as an adjacency list
    graph = {}
    for index, row in edges.iterrows():
        source = row['source']
        target = row['target']
        weight = row['weight']
        if source not in graph:
            graph[source] = []
        graph[source].append((target, weight))

    # Initialize distances and predecessors for all nodes in the graph
    all_node_ids = set(nodes['node_id'].tolist()) | set(edges['source'].tolist()) | set(edges['target'].tolist())
    distances = {node_id: float('infinity') for node_id in all_node_ids}
    distances[start_node_id] = 0
    predecessors = {node_id: None for node_id in all_node_ids}

    # Priority queue for Dijkstra
    priority_queue = [(0, start_node_id)]

    while priority_queue:
        current_distance, current_node = heapq.heappop(priority_queue)

        # If the current distance is greater than the known distance, skip
        if current_distance > distances[current_node]:
            continue

        # Explore neighbors
        if current_node in graph:
            for neighbor, weight in graph[current_node]:
                distance = current_distance + weight

                # If a shorter path is found
                if distance < distances[neighbor]:
                    distances[neighbor] = distance
                    predecessors[neighbor] = current_node
                    heapq.heappush(priority_queue, (distance, neighbor))

    # Filter out unreachable nodes
    reachable_distances = {node_id: dist for node_id, dist in distances.items() if dist != float('infinity')}
    reachable_predecessors = {node_id: pred for node_id, pred in predecessors.items() if pred is not None or node_id == start_node_id} # Include start node with None predecessor

    return reachable_distances, reachable_predecessors
start_node = 1398
# Example usage: Find the shortest path from start_node to all reachable nodes
shortest_paths, path_predecessors = dijkstra_shortest_path(edges_all, nodes_all, start_node)

print(f"\nShortest paths from node {start_node} to all reachable nodes:")
if shortest_paths:
    # Sort by distance for better readability
    sorted_paths = sorted(shortest_paths.items(), key=lambda item: item[1])
    for target_node_id, distance in sorted_paths:
        # Reconstruct the path
        path = []
        current = target_node_id
        # Handle the case where the target is the start node itself
        if current == start_node:
            path = [start_node]
        else:
            while current is not None:
                path.append(current)
                current = path_predecessors.get(current)
            path.reverse()
        print(f"  To node {target_node_id}: Distance = {distance}, Path = {path}")
else:
    print(f"  No reachable nodes found from node {start_node}.")
```
